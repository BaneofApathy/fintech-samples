An investigator wants to see which accounts share devices, merchants, or addresses. A graph draws each account or shared object as a point and each recorded relationship as a line. These links help organize evidence; they do not establish that an account is fraudulent or that money moved along every line.

**Dataset:** This teaching graph is written by hand: four account points, two device points, one merchant point, one address point, and seven relationship lines. It has no transaction outcomes, dates, or fraud labels, so it can show connections but cannot measure fraud-detection performance.

**Task:** Find the connected groups, count how many links touch each point, and count the shared neighbors of accounts A and B. PageRank is a score that gives more weight to links with well-connected points. Explain what each result says about the recorded network and what it cannot say about wrongdoing.

**Compare:** Remove the recorded line between account A and merchant X, then compare the A/B shared-neighbor count and the connected groups. Only that line changes. In a historical test, also filter every link by its recorded time: a link first recorded after the decision cutoff was not available to that decision. The exercise shows how a network summary responds to one relationship; it does not tell us whether the removed relationship was true or useful in a real investigation.

**Runtime:** NetworkX stores and measures this small graph. The graph has no timestamps, so the code cannot tell whether a relationship was already known at the time a past decision was made. In a real evaluation, filter links by their recorded time before building the graph; otherwise later information can make an earlier decision look better than it was.


````python
import networkx as nx
G = nx.Graph()
edges = [("acct_A","device_7"),("acct_B","device_7"),
    ("acct_A","merchant_X"),("acct_B","merchant_X"),
    ("acct_B","address_4"),("acct_C","address_4"),("acct_D","device_99")]
G.add_edges_from(edges)
components = sorted(nx.connected_components(G),key=len,reverse=True)
print("Largest component:",sorted(components[0]))
print("Degrees:",sorted(G.degree,key=lambda x:x[1],reverse=True))
print("A/B common neighbors:",len(list(nx.common_neighbors(G,"acct_A","acct_B"))))
print("PageRank:",{k:round(v,3) for k,v in nx.pagerank(G).items()})
print("Baseline: account-only graph has no edges and therefore no shared-device evidence.")
````
