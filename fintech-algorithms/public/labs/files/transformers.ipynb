Label each short financial sentence as positive, neutral, or negative, then inspect where wording changes the meaning. The browser uses a small word-count model called TF-IDF with logistic regression. It is a comparison model, not FinBERT; the separate native download runs FinBERT.

**Dataset:** There are 12 hand-written training sentences (4 for each label) and 6 labeled test sentences. The test set is tiny and hand-made, so its scores are practice numbers, not evidence that the model understands finance in general.

**Task:** Fit the word-based comparison model, compare each prediction with its supplied label, and report macro-F1 plus the per-class report. Macro-F1 averages the F1 score for positive, neutral, and negative separately, so a larger class does not drown out errors on a smaller class.

**Compare:** Keep the six test sentences fixed. First inspect the errors, then strengthen regularization or correct the negation example and rerun. Regularization discourages the model from relying too heavily on individual word patterns. A score change on six sentences is a debugging clue, not proof that the change will work on future documents.

**Runtime:** The browser notebook uses scikit-learn: TF-IDF turns words into weighted counts and logistic regression maps those counts to one of the three labels. It does not download or run FinBERT. The separate native example uses the pretrained FinBERT language model and needs transformers, PyTorch, and an initial model download.

[Download the native Python example](native/transformers.py).


````python
# Comparison model: TF-IDF weights word counts; logistic regression selects one label.
# This is not FinBERT. The separate native/transformers.py download runs FinBERT.
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, classification_report
train = ["profit increased", "margin expanded", "revenue rose", "growth improved",
 "profit fell", "guidance withdrawn", "revenue declined", "loss widened",
 "scheduled debt payment", "report published", "meeting held", "filing submitted"]
labels = ["positive"]*4+["negative"]*4+["neutral"]*4
test_cases = [("profit rose", "positive"), ("margin improved", "positive"),
 ("guidance declined", "negative"), ("loss fell", "positive"),
 ("scheduled report published", "neutral"), ("meeting submitted", "neutral")]
test, truth = zip(*test_cases)
# C controls regularization: a smaller value discourages reliance on individual word patterns.
model = make_pipeline(TfidfVectorizer(),LogisticRegression(C=1.0,max_iter=500))
model.fit(train,labels)
pred = model.predict(test)
print("Macro-F1:",f1_score(truth,pred,average="macro"))
print(classification_report(truth,pred,zero_division=0))
print("Challenge:","loss fell", "prediction:",pred[3])
print("This tiny sample is for debugging, not evidence of finance-domain accuracy.")
````
